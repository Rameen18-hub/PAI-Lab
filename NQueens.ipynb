```python
def is_safe(board, row, col, N):
    for i in range(col):
        if board[row][i] == 1:
            return False
        
    i, j = row,col
    while i >=0 and j >= 0:
        if board[i][j] ==1:
            return False
        i -= 1
        j -= 1

    i,j = row, col

    while i < N and j >= 0:
        if board[i][j] ==1:
            return False
        i+=1
        j-=1

    return True

def solve_n_queens(board, col, N, solutions, steps):
    if col >=N:
        solutions.append([row[:] for row in board])
        return True
    
    for row in range(N):
        steps.append(f"Trying row {row}, col {col}")
        if is_safe(board, row, col, N):
            board[row][col] = 1
            steps.append(f"PLaced quen at({row}, {col})")
            solve_n_queens(board, col + 1, N, solutions, steps)

            board[row][col] = 0
            steps.append(f"Backtyracking from ({row}, {col})")

    return False

def n_queens(N):
    board = [[0 for _ in range(N)] for _ in range(N)]
    solutions =[]
    steps=[]

    solve_n_queens(board, 0, N, solutions,steps)

    print("Steps taken:")
    for s in steps:
        print(s)

    print(f"\nTotal solutions forN={N}: {len(solutions)}")
    for idx, sol in enumerate(solutions, start=1):
        print(f"\nSolution {idx}:")
        for row in sol:
            print(row)

n_queens(4)
```
