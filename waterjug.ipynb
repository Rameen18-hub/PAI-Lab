```python
CAP_A =4
CAP_B =3
GOAL =2

def get_neighbours(state):
    x,y=state
    neighbours=[]
    neighbours.append(((CAP_A, y),"Fill Jug A"))
    neighbours.append(((x, CAP_B),"Fill Jug B"))
    neighbours.append(((0, y),"Empty Jug A"))
    neighbours.append(((x, 0),"Empty Jug B"))

    pour = min(x,CAP_B - y)
    neighbours.append(((x - pour, y + pour), "Pour Jug A --> Jug B"))
    pour = min(y,CAP_B - x)
    neighbours.append(((x + pour, y - pour), "Pour Jug B --> Jug A"))

    return neighbours

def dfs_water_jug(state=(0,0), visited=None, path= None, rules= None):
    if visited is None: visited = set()
    if path is None: path =[]
    if rules is None: rules = []

    visited.add(state)
    path.append(state)

    if state[0] == GOAL or state [1] == GOAL:
        return path,rules
    
    for neighbor, rule in get_neighbours(state):
        if neighbor not in visited:
            new_path, new_rules = dfs_water_jug(
                neighbor,visited.copy(),path.copy(),rules.copy() + [rule]
            )
            if new_path:
                return new_path,new_rules
            
    return None,None

path,rules = dfs_water_jug()

print("DFS Water Jug Solution Path:")
for i, p in enumerate(path):
    if i < len (rules):
        print(f"Step {i+1}: {rules[i]} --> State:{path[i+1]}")

print("Total steps:" , len(rules))

```
